```
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from google.colab import files

# =========================================================
# TAHAP 1: INPUT INTERAKTIF & CRAWLING
# =========================================================
print("=== SISTEM MONITORING TREN INDUSTRI (NAIVE BAYES) ===")
url_input = input("Masukkan url IndustryWeek : ")

def ambil_berita(url):
    headers = {'User-Agent': 'Mozilla/5.0'}
    try:
        response = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(response.text, 'html.parser')
        # Mencari judul berita dalam tag h2 atau h3
        titles = [a.get_text(strip=True) for a in soup.find_all(['h2', 'h3'], limit=12)]
        if not titles: raise Exception
        return titles
    except:
        print("\n[!] Gagal crawling atau URL tidak merespon. Menggunakan data simulasi untuk demo...")
        return [
            "Smart Factory Automation with AI",
            "Global Supply Chain Logistics Disruptions",
            "Workforce Skills Training and Labor Shortage",
            "Digital Twins for Production Optimization",
            "Freight Shipping Costs Increase in 2024",
            "Employee Safety and Industrial HSE Standards"
        ]

data_berita = ambil_berita(url_input)

# =========================================================
# TAHAP 2: TRAINING MODEL (KNOWLEDGE BASE)
# =========================================================
data_training = [
    "artificial intelligence robotics automation smart factory iot digital twin",
    "bathroom holidays logistics supply chain freight shipping warehouse transportation procurement",
    "human resources safety labor skills training employee recruitment hse"
]
labels = ["Digital Transformation", "Supply Chain & Logistics", "Operational & HSE"]

# =========================================================
# TAHAP 3: IMPLEMENTASI BoW, TF-IDF, & NAIVE BAYES
# =========================================================

# 1. Proses Bag of Words (BoW)
bow_vec = CountVectorizer(stop_words='english')
bow_matrix = bow_vec.fit_transform(data_berita)
df_bow = pd.DataFrame(bow_matrix.toarray(), columns=bow_vec.get_feature_names_out())

# 2. Proses TF-IDF
tfidf_vec = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vec.fit_transform(data_berita)
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vec.get_feature_names_out())

# 3. Klasifikasi Naive Bayes
# Kita gunakan data training untuk melatih model
X_train = tfidf_vec.fit_transform(data_training)
model_nb = MultinomialNB()
model_nb.fit(X_train, labels)

# Prediksi data hasil crawling
X_test = tfidf_vec.transform(data_berita)
hasil_prediksi = model_nb.predict(X_test)

# =========================================================
# TAHAP 4: OUTPUT LAPORAN (UNTUK PRESENTASI)
# =========================================================
print("\n" + "="*50)
print("HASIL EKSTRAKSI BAG OF WORDS (BoW)")
print("="*50)
display(df_bow.head(10))

print("\n" + "="*50)
print("HASIL PEMBOBOTAN TF-IDF")
print("="*50)
display(df_tfidf.head(10))

print("\n" + "="*50)
print("HASIL KLASIFIKASI METODE NAIVE BAYES")
print("="*50)
df_final = pd.DataFrame({
    'Judul Berita': data_berita,
    'Kategori Klasifikasi': hasil_prediksi
})
display(df_final)

# Download Hasil ke CSV
df_final.to_csv("Hasil_Analisis_Industri.csv", index=False)
files.download("Hasil_Analisis_Industri.csv")
```

## Laporan Analisis Tren Industri Berdasarkan Berita IndustryWeek

### Pendahuluan
Dokumen ini menyajikan hasil analisis tren industri menggunakan data berita dari IndustryWeek. Analisis dilakukan dengan metode Bag of Words (BoW), TF-IDF, dan klasifikasi Naive Bayes untuk mengkategorikan berita ke dalam topik-topik relevan seperti 'Digital Transformation', 'Supply Chain & Logistics', dan 'Operational & HSE'.

### Metode Analisis
1.  **Pengambilan Data**: Berita diambil dari URL IndustryWeek yang ditentukan.
2.  **Preprocessing Teks**: Judul berita diproses untuk menghasilkan representasi numerik menggunakan:
    *   **Bag of Words (BoW)**: Menghitung frekuensi kemunculan kata.
    *   **TF-IDF**: Memberikan bobot pada kata berdasarkan relevansinya.
3.  **Klasifikasi Naive Bayes**: Model dilatih menggunakan data training yang telah ditentukan untuk memprediksi kategori berita.
